# County Rainfall and Planting Advisory

## Project Summary

This project analyses monthly rainfall data for five Kenyan counties: Nairobi, Nakuru, Kisumu, Mombasa, and Machakos. Python and NumPy are used to clean rainfall data, calculate annual rainfall totals and monthly averages, identify wettest months and dry months, and recommend planting windows. The results will help a farmer-support NGO provide rainfall-based planting advice to farmers.


```python

import numpy as np

# List of months
months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

# Raw Machakos rainfall data
machakos_raw = [
    "48", "35", "88", "N/A", "140", " 20 ",
    "9", "12", "", "45", "182", "96"
]

# Function to clean Machakos rainfall data
def clean_machakos(raw_values):
    valid_values = []

    for value in raw_values:
        try:
            valid_values.append(int(value.strip()))
        except ValueError:
            pass

    average = round(np.mean(valid_values))

    cleaned_values = []
    imputed_months = []

    for month, value in zip(months, raw_values):
        try:
            cleaned_values.append(int(value.strip()))
        except ValueError:
            cleaned_values.append(average)
            imputed_months.append(month)

    return cleaned_values, imputed_months, average


# Clean the data
machakos_clean, imputed_months, imputed_value = clean_machakos(machakos_raw)

print("Machakos Cleaning Results")
print("Imputed rainfall value:", imputed_value, "mm")
print("Months imputed:", imputed_months)
print("Cleaned Machakos data:", machakos_clean)
```

```python

# List of counties
counties = ["Nairobi", "Nakuru", "Kisumu", "Mombasa", "Machakos"]

# Rainfall data for the first four counties
clean_rows = [
    [64, 56, 92, 219, 176, 35, 18, 24, 31, 61, 165, 104],
    [35, 42, 70, 145, 120, 78, 96, 105, 72, 55, 60, 38],
    [58, 90, 150, 240, 180, 95, 70, 85, 90, 88, 130, 95],
    [30, 15, 55, 150, 240, 110, 90, 65, 60, 90, 100, 70]
]

# Add the cleaned Machakos data from Step 4
all_rows = clean_rows + [machakos_clean]

# Create a NumPy array
R = np.array(all_rows, dtype=int)

# Display the results
print("Combined Rainfall Array:")
print(R)

print("\nArray shape:", R.shape)
print("Array data type:", R.dtype)
```

```python

# Calculate annual rainfall totals for each county
annual_totals = np.sum(R, axis=1)

# Display the annual totals
print("ANNUAL RAINFALL TOTALS")

for county, total in zip(counties, annual_totals):
    print(f"{county}: {total} mm")
```

```python

# Calculate average rainfall for each month
monthly_averages = np.mean(R, axis=0)

# Display the monthly averages
print("AVERAGE MONTHLY RAINFALL ACROSS FIVE COUNTIES")

for month, average in zip(months, monthly_averages):
    print(f"{month}: {average:.2f} mm")
```

```python

# Find the index of the wettest month for each county
wettest_indices = np.argmax(R, axis=1)

# Convert the indices into month names
wettest_months = [months[i] for i in wettest_indices]

# Display the wettest month for each county
print("WETTEST MONTH PER COUNTY")

for county, month in zip(counties, wettest_months):
    print(f"{county}: {month}")
```

```python

# Count months with rainfall below 50 mm
dry_months_count = np.sum(R < 50, axis=1)

# Display the number of dry months per county
print("NUMBER OF DRY MONTHS PER COUNTY")

for county, count in zip(counties, dry_months_count):
    print(f"{county}: {count} dry months")
```

```python

# Add rainfall from each pair of consecutive months
consecutive_totals = R[:, :-1] + R[:, 1:]

# Find the position of the highest combined rainfall
best_window_indices = np.argmax(consecutive_totals, axis=1)

# Create the recommended planting windows
planting_windows = [
    f"{months[i]}-{months[i + 1]}"
    for i in best_window_indices
]

# Display the results
print("RECOMMENDED PLANTING WINDOWS")

for county, window in zip(counties, planting_windows):
    print(f"{county}: {window}")
```

```python

import csv

# Create the rainfall summary CSV file
with open("rainfall_summary.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    # Write the column headings
    writer.writerow([
        "county",
        "annual_total_mm",
        "wettest_month",
        "dry_months",
        "planting_window"
    ])

    # Write one row for each county
    for county, total, wettest, dry_count, window in zip(
        counties,
        annual_totals,
        wettest_months,
        dry_months_count,
        planting_windows
    ):
        writer.writerow([
            county,
            int(total),
            wettest,
            int(dry_count),
            window
        ])

print("rainfall_summary.csv created successfully!")
```
